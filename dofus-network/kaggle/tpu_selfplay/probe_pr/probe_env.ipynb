```python
import glob, os, subprocess, sys
print('accel devices', glob.glob('/dev/accel*'))
print('tpu env', {k: v for k, v in os.environ.items() if 'TPU' in k.upper()})
try:
    import tensorflow as tf
    print('tf', tf.__version__, tf.config.list_physical_devices())
except Exception as exc:
    print('tf error', exc)
code = "import jax; print('backend', jax.default_backend()); print('devices', jax.devices()); print('count', jax.local_device_count())"
for acc in ('v5litepod-8', 'v5e-8'):
    env = dict(os.environ, TPU_ACCELERATOR_TYPE=acc, TPU_NAME='local')
    try:
        r = subprocess.run([sys.executable, '-c', code], capture_output=True, text=True, timeout=120, env=env)
        print('ACC', acc, 'rc', r.returncode)
        print(r.stdout[-2000:])
        print(r.stderr[-2000:])
    except Exception as exc:
        print('ACC', acc, 'error', exc)
```
